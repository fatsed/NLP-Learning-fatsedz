```python
import re, string
# 1 - Load & Clean Text
with open("/content/myText.txt","r",encoding="utf-8") as f:
    raw = f.read()

text1 = "".join(ch for ch in raw if ch not in string.punctuation)
text2 = text1.replace("\n"," ")
clean_text = re.sub(r"\s+", " ", text2).strip()

clean_text[:300]
```

```python
import nltk
nltk.download('all')
```

```python
# 2- Tokenize + Remove Stopwords
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

tokens = word_tokenize(clean_text)
stop = set(stopwords.words("english"))
tokens_no_stop = [w for w in tokens if w.lower() not in stop]

tokens_no_stop[:30]
```

```python
# 3- Stemming + Lemmatization
from nltk.stem import PorterStemmer, WordNetLemmatizer

porter = PorterStemmer()
lemma = WordNetLemmatizer()

stemmed = [porter.stem(w) for w in tokens_no_stop]
lemmatized = [lemma.lemmatize(w) for w in tokens_no_stop]

list(zip(tokens_no_stop[:20], stemmed[:20], lemmatized[:20]))
```

```python
# 4- N-grams (unigram, bigram, trigram)
from nltk import ngrams
from nltk.tokenize import WordPunctTokenizer

wp = WordPunctTokenizer()
toks = wp.tokenize(clean_text)

bigrams = list(ngrams(toks, 2))
trigrams = list(ngrams(toks, 3))

bigrams[:10], trigrams[:10]
```

```python
# 5 - POS Tagging
pos_tags = nltk.pos_tag(tokens_no_stop)
pos_tags[:20]
```

```python
# 6 - NER – Named Entity Recognition
entities = []
for sent in nltk.sent_tokenize(clean_text):
    tree = nltk.ne_chunk(nltk.pos_tag(word_tokenize(sent)))
    for chunk in tree:
        if hasattr(chunk, 'label'):
            entities.append((chunk.label(), " ".join(c[0] for c in chunk)))
entities[:20]
```

```python
# 7 - Vectorization – CountVectorizer + Keras Tokenizer
from sklearn.feature_extraction.text import CountVectorizer
from tensorflow import keras

# split text into mini-documents
docs = [s for s in nltk.sent_tokenize(clean_text) if len(s.split()) > 3]

# Keras Tokenizer
tokenizer = keras.preprocessing.text.Tokenizer()
tokenizer.fit_on_texts(docs)
print("Vocabulary:", list(tokenizer.word_index.keys()))
matrix = tokenizer.texts_to_matrix(docs, mode="count")
print("\nKeras count matrix:")
print(matrix)

# CountVectorizer
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(docs)
print("\nCountVectorizer vocabulary:")
print(vectorizer.get_feature_names_out())
print("\nCountVectorizer matrix (dense):")
print(X.toarray())

X.shape, matrix.shape
```
