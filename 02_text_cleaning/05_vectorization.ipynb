```python
from tensorflow import keras
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "I love natural language processing",
    "Text cleaning is the first step in NLP",
    "I love cleaning and processing text data"
]

# 1) Keras Tokenizer
tokenizer = keras.preprocessing.text.Tokenizer()
tokenizer.fit_on_texts(corpus)
print("Vocabulary:", list(tokenizer.word_index.keys()))

matrix = tokenizer.texts_to_matrix(corpus, mode="count")
print("\nKeras count matrix:")
print(matrix)

# 2) CountVectorizer
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

print("\nCountVectorizer vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nCountVectorizer matrix (dense):")
print(X.toarray())
```
