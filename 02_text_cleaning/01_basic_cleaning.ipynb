```python
import string
import re

#read raw text
with open('/content/myText.txt', 'r', encoding='utf_8') as f:
  text = f.read()
print("sample raw text:")
print(text[:300])
print("-"*50)
print("Length before cleaning", len(text))

#remove punctuation
str_pun = string.punctuation
text_c1 = "".join(ch for ch in text if ch not in str_pun)
print("Length after removing punctuation:", len(text_c1))

#remove newlines
text_c2 = text_c1.replace('\n', " ")

#remove digits + normalize whitespace
text_c3 = re.sub(r"[0-9]", "", text_c2)  # remove all digits
text_c3 = re.sub(r"\s+", " ", text_c3)   # collapse spaces/newlines/tabs to 1 space
print("-"*50)
print("Length after removing digits & normalizing spaces:", len(text_c3))
print("Cleaned sample:")
print(text_c3[:300])
```
