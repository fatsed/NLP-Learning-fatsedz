```python
from nltk import ngrams
from nltk.tokenize import WordPunctTokenizer
text = "How much wood would a woodchuck chuck if a woodchuck could chuck wood?"
tokens = WordPunctTokenizer().tokenize(text)
print("Tokens:", tokens)

# bigrams
bigrams = [tuple(w) for w in ngrams(tokens, n=2)]
print("\nBigrams:")
print(bigrams)

# trigrams
trigrams = ["_".join(w) for w in ngrams(tokens, n=3)]
print("\nTrigrams (joined with _):")
print(trigrams)
```

```python
from pathlib import Path
import string, re
import nltk

with open("/content/myText.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()

text_c1 = "".join(ch for ch in raw_text if ch not in string.punctuation)
text_c2 = text_c1.replace("\n", " ")
text_c3 = re.sub(r"\s+", " ", text_c2)
#Named Entity Recognition (NER)
for sent in nltk.sent_tokenize(text_c3):
    for chunk in nltk.ne_chunk(nltk.pos_tag(nltk.word_tokenize(sent))):
        if hasattr(chunk, "label"):
            print(f"{chunk.label()} --> ", " ".join(c[0] for c in chunk))
```
