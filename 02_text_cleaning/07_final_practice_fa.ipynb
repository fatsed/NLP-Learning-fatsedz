```python
from hazm import Normalizer, word_tokenize, Stemmer, Lemmatizer, sent_tokenize, POSTagger
from hazm.utils import stopwords_list
import re
import string
```

```python
# 2 - Text Cleaning + Normalizer (Hazm)
normalizer = Normalizer()
def clean_text(text):

  # remove English letters
  text = normalizer.normalize(text)
  # remove digits (English + Persian)
  text = re.sub(r"[A-Za-z]", " ", text)
  clean_text = re.sub(r'[^\u0600-\u06FF\s]', ' ', text)
  text = re.sub(r"[0-9۰-۹]", " ", text)
  # remove extra punctuation
  punct = string.punctuation + "«»؟؛،ـ…“”"
  text = "".join(ch for ch in text if ch not in punct)
  # collapse multiple whitespaces
  text = re.sub(r"\s+", " ", text)
  return text

cleaned_text = clean_text(raw_text)
print("Cleaned text sample:\n")
print(cleaned_text[:800])
```

```python
# 3 - Tokenization + stopwords
# sentence tokenize
sentences = sent_tokenize(cleaned_text)
print("Number of sentences:", len(sentences))
print("Sample sentence:", sentences[0])

# word tokenize
tokens = word_tokenize(cleaned_text)
print("\nNumber of tokens:", len(tokens))
print("Sample tokens:", tokens[:40])

#stopwords
fa_stopwords = set(stopwords_list())
print("\nNumber of Persian stopwords:", len(fa_stopwords))

tokens_no_stop = [w for w in tokens if w not in fa_stopwords]
print("Number of tokens after stopword removal:", len(tokens_no_stop))
print("Sample tokens (no stopwords):", tokens_no_stop[:40])
```

```python
# 4 - Stemming & Lemmatization
stemmer = Stemmer()
lemmatizer = Lemmatizer()

sample_tokens = tokens_no_stop[:60]

rows = []
for w in sample_tokens:
    rows.append((w, stemmer.stem(w), lemmatizer.lemmatize(w)))

print("word | stem | lemma")
print("-" * 40)
for w, s, l in rows:
    print(f"{w:15s} | {s:15s} | {l:15s}")
```

```python
# 5 - N-grams
from collections import Counter
from nltk.util import ngrams

def top_ngrams(tokens, n=2, top_k=15):
    ng = ngrams(tokens, n)
    counts = Counter(ng)
    return counts.most_common(top_k)

top_unigrams = top_ngrams(tokens_no_stop, n=1)
top_bigrams  = top_ngrams(tokens_no_stop, n=2)
top_trigrams = top_ngrams(tokens_no_stop, n=3)

print("Top unigrams:")
for ng, c in top_unigrams:
    print(c, ":", " ".join(ng))

print("\nTop bigrams:")
for ng, c in top_bigrams:
    print(c, ":", " ".join(ng))

print("\nTop trigrams:")
for ng, c in top_trigrams:
    print(c, ":", " ".join(ng))
```

```python
# 7 - Vectorization (CountVectorizer + Keras)
from sklearn.feature_extraction.text import CountVectorizer
from tensorflow import keras

# Use Persian sentences as documents
docs_fa = [s for s in sentences if len(s.split()) > 3]
print("Number of docs:", len(docs_fa))
print("Sample doc:", docs_fa[0])

# CountVectorizer with hazm tokenizer
def hazm_tokenizer(text):
    text = clean_text(text)
    return [w for w in word_tokenize(text) if w not in fa_stopwords]


vectorizer = CountVectorizer(tokenizer=hazm_tokenizer)
X = vectorizer.fit_transform(docs_fa)
print("CountVectorizer vocab size:\n", len(vectorizer.get_feature_names_out()))
print("Matrix shape:", X.shape)
print("First 20 vocab items:\n", vectorizer.get_feature_names_out()[:20])
print("\nCountVectorizer matrix (dense):")
print(X.toarray())

# Keras Tokenizer on Persian docs
tokenizer_keras = keras.preprocessing.text.Tokenizer()
tokenizer_keras.fit_on_texts(docs_fa)
print("Keras vocab size:", len(tokenizer_keras.word_index))
print("First 20 vocab items:", list(tokenizer_keras.word_index.keys())[:20])
matrix_keras = tokenizer_keras.texts_to_matrix(docs_fa, mode="count")
print("Keras matrix shape:", matrix_keras.shape)
```
