```python
import requests as rq
from bs4 import BeautifulSoup
import re
url = "https://trtfarsi.com"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) Chrome/118.0.0.0 Safari/537.36"
}
res = rq.get(url, headers = headers)
res.encoding = 'utf-8'
if res.status_code == 200:
  soup = BeautifulSoup(res.text, "lxml")
  for tag in soup(["script", "style"]):
    tag.decompose()
  text_raw = soup.get_text(separator=" ")
  text_raw = re.sub(r"\s+", " ", text_raw).strip()
  print("📄 Length of raw text:", len(text_raw))
  print("📌 Sample raw text:\n", text_raw[:800])
else:
    print("eroorrr!!!! can't access to file")
```

```python
from hazm import Normalizer, word_tokenize
from hazm.utils import stopwords_list
import re

normalizer = Normalizer()
fa_stopwords = set(stopwords_list())

def clean_persian_text(text: str) -> str:
    text = normalizer.normalize(text)
    text = re.sub(r"[0-9۰-۹]+", " ", text)
    text = re.sub(r"[A-Za-z]+", " ", text)
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

text_raw_norm = normalizer.normalize(text_raw)

text_clean = clean_persian_text(text_raw)

print("طول متن خام نرمال:", len(text_raw_norm))
print("طول متن تمیز:", len(text_clean))
print("\n نمونه متن تمیز:\n", text_clean[:500])
```

```python
tokens_raw = word_tokenize(text_raw_norm)
tokens_clean = word_tokenize(text_clean)

tokens_raw_nostop = [w for w in tokens_raw if w not in fa_stopwords and len(w) > 1]
tokens_clean_nostop = [w for w in tokens_clean if w not in fa_stopwords and len(w) > 1]

print("تعداد توکن‌های خام:", len(tokens_raw))
print("بعد از حذف استاپ‌ورد (خام):", len(tokens_raw_nostop))
print("تعداد توکن‌های تمیز:", len(tokens_clean))
print("بعد از حذف استاپ‌ورد (تمیز):", len(tokens_clean_nostop))

print("\nچند توکن خام:", tokens_raw_nostop[:30])
print("چند توکن تمیز:", tokens_clean_nostop[:30])
```

```python
from collections import Counter

counter_raw = Counter(tokens_raw_nostop)
counter_clean = Counter(tokens_clean_nostop)

top10_raw = counter_raw.most_common(10)
top10_clean = counter_clean.most_common(10)

print("🔹 Top 10 words in RAW text (no stopwords):")
for w, c in top10_raw:
    print(f"{w:15}  ->  {c}")

print("\n🔹 Top 10 words in CLEAN text (no stopwords):")
for w, c in top10_clean:
    print(f"{w:15}  ->  {c}")
```

```python
import pandas as pd
df_freq = pd.DataFrame({
    "raw_word": [w for w, _ in top10_raw],
    "raw_count": [c for _, c in top10_raw],
    "clean_word": [w for w, _ in top10_clean],
    "clean_count": [c for _, c in top10_clean],
})

df_freq
```
