```python
import requests as rq
import json

url = "http://universities.hipolabs.com/search"
payload = {
    "country": "United States",
    "name": "tech"
}
res = rq.get(url, params=payload)
if res.status_code == 200:
    print("Final URL:", res.url)
    data = res.json()
    print("Number of universities found:", len(data))
```

```python
print(data[0])
```

```python
print("\nSample results:")
for uni in data[:5]:
        print("-" * 40)
        print("Name      :", uni.get("name"))
        print("Country   :", uni.get("country"))
        print("City      :", uni.get("state-province"))
        print("Website(s):", ", ".join(uni.get("web_pages", [])))
else:
    print("Request failed with status:", res.status_code)
```
