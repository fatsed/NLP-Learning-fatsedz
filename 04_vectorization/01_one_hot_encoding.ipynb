```python
text = ['red', 'yellow', 'green', 'red']
# Build vocabulary (unique sorted words)
# set -> Removes duplicate words | Converts back to list | Sorts alphabetically
vocab = sorted(list(set(text)))
print("Vocabulary:", vocab)

# Map word -> index
#creates a dictionary: -> Key = word, value = column number
#{"green": 0, "red": 1, "yellow": 2}
word_to_index = {word: i for i, word in enumerate(vocab)}
print("word_to_index:", word_to_index)

def one_hot(word):
    """Return one-hot vector for a given word in the vocabulary."""
    vector = [0] * len(vocab)  # [0, 0, 0]
    vector[word_to_index[word]] = 1
    return vector

for w in text:
    print(f"{w:6} -> {one_hot(w)}")
```

```python
sentence = "I like python language programming"
tokens = sentence.split()
vocab = sorted(list(set(tokens)))
word_to_index = {word: i for i, word in enumerate(vocab)}

def one_hot(word):
    """Return one-hot vector for a given word in the vocabulary."""
    vector = [0] * len(vocab)  # [0, 0, 0]
    vector[word_to_index[word]] = 1
    return vector

def sentence_to_matrix(tokens):
  return [one_hot(w) for w in tokens]
matrix = sentence_to_matrix(tokens)

print("Tokens:", tokens)
print("One-hot matrix:")
for row in matrix:
    print(row)
```
