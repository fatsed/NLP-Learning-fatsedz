```python
model = api.load("word2vec-google-news-300")

print("Model loaded ✅")
print("Vocab size:", len(model.key_to_index))
print("Vector size:", model.vector_size)
```

```python
# Inspect Single Word Vectors
vec_book = model["book"]
print("Vector shape:", vec_book.shape)
print("First 10 components:", vec_book[:10])
```

```python
 #Finding Most Similar Words
for w in ["book", "apple", "university", "king", "beautiful"]:
  print(f"\nTop 5 similar words to '{w}':")
  for word, score in model.most_similar(w, topn=5):
    print(f"  {word:<15}  {score:.3f}")
```

```python
#Word Analogies
result = model.most_similar(
    positive=["king", "woman"],
    negative=["man"],
    topn=5
)

print("Analogy: king - man + woman ≈ ?")
for word, score in result:
    print(f"{word:<15} {score:.3f}")

print("-"*50)

coun_res = model.most_similar(
    positive=["paris", "italy"],
    negative=["france"],
    topn=5
)
print("paris - france + italy ≈ ?")
for word, score in coun_res:
    print(f"{word:<15} {score:.3f}")

print("-"*50)

wo_res = model.most_similar(
    positive=["walking", "swim"],
    negative=["walk"],
    topn=5
)
print("walking - walk + swim ≈ ?")
for word, score in wo_res:
    print(f"{word:<15} {score:.3f}")

print("-"*50)

siz_res = model.most_similar(
    positive=["big", "bigger"],
    negative=["smaller"],
    topn=5
)
print("big - bigger + smaller ≈ ?")
for word, score in siz_res:
    print(f"{word:<15} {score:.3f}")
```

```python
#Cosine Similarity Between Words
pairs = [
    ("apple", "banana"),
    ("apple", "cat"),
    ("cat", "dog"),
    ("king", "queen"),
    ("man", "woman")
]

for w1, w2 in pairs:
    sim = model.similarity(w1, w2)
    print(f"similarity({w1:>8}, {w2:<8}) = {sim:.3f}")
```

```python
#Similarity Matrix for a Set of Words
import numpy as np
import pandas as pd

words = ["king", "queen", "man", "woman", "prince", "princess"]

# Build similarity matrix
sim_matrix = np.zeros((len(words), len(words)))

for i, w1 in enumerate(words):
    for j, w2 in enumerate(words):
        sim_matrix[i, j] = model.similarity(w1, w2)

sim_df = pd.DataFrame(sim_matrix, index=words, columns=words)
sim_df
```

```python
# Filtering Vocabulary by Prefix
prefix = "covid"

vocab = sorted([
    w for w in model.key_to_index.keys()
    if w.lower().startswith(prefix)
])

print(f"Found {len(vocab)} words starting with '{prefix}':")
print(vocab[:50])
```

```python
#Handling Out-of-Vocabulary (OOV) Words
def safe_most_similar(word, topn=5):
    if word in model.key_to_index:
        return model.most_similar(word, topn=topn)
    else:
        print(f"'{word}' is OOV (not in vocabulary).")
        return []

safe_most_similar("Alexis")
safe_most_similar("Iran")
```

```python
#Fuzzy Matching
from difflib import get_close_matches

def suggest_similar_spelling(word, n=5, cutoff=0.7):
    candidates = get_close_matches(word, model.key_to_index.keys(), n=n, cutoff=cutoff)
    print(f"Closest matches to '{word}':", candidates)

suggest_similar_spelling("Alexis")
suggest_similar_spelling("Germani")
```

```python
import numpy as np

# get list of vocab words
vocab_list = list(model.key_to_index.keys())

# sample 5 random words
np.random.seed(42)
random_words = np.random.choice(vocab_list, 5)
print("Random words:", random_words)

for w in random_words:
    print(f"\nNeighbors of '{w}':")
    try:
        for word, score in model.most_similar(w, topn=5):
            print(f"  {word:<15} {score:.3f}")
    except KeyError:
        print("  (Not in vocab)")
```
