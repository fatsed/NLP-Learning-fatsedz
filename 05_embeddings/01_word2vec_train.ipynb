```python
sentences = [
    ["i", "love", "machine", "learning"],
    ["i", "love", "data", "science"],
    ["word", "embeddings", "are", "amazing"]
]
```

```python
from gensim.models import Word2Vec

model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1  # skip-gram
)

model.save("my_word2vec.model")
```

```python
model.wv.most_similar("love")
```
