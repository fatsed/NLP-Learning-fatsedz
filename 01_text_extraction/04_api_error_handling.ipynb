```python
import requests as rq
import time
```

```python
response = rq.get(" https://api.chucknorris.io/jokes/random")
if response.status_code == 429:
  retry_after = int(response.headers.get("Retry-After", 1))
  print(f"Rate limit exceeded. Waiting {retry_after} seconds before retrying...")
  time.sleep(retry_after)
  response = rq.get('https://example.com/api')
```

```python
# 3 time try
url = "https://api.adviceslip.com/advice"

for attempt in range(3):
  r=rq.get(url)

  if r.status_code == 200:
    print("Success:", r.json())
    break
  elif r.status_code == 429:
    retry_after = int(response.headers.get("Retry-After", 1))
    print(f"Rate limit exceeded. Waiting {retry_after} seconds before retrying...")
    time.sleep(retry_after)
  else:
    print("Other error:", r.status_code)
    break
```

```python
import requests
import time

url = "https://httpbin.org/status/429"

response = requests.get(url)

if response.status_code == 429:
    retry_after = 3
    print(f"Rate limit exceeded. Waiting {retry_after} seconds...")
    time.sleep(retry_after)
    response = requests.get("https://api.adviceslip.com/advice")

print("Final status:", response.status_code)
print("Data:", response.text)
```
