```python
import urllib3
http = urllib3.PoolManager()
url ="https://catfact.ninja/fact"
headers = urllib3.util.make_headers(user_agent="fatsedz/1.0.1", basic_auth="abc:xyz")
```

```python
import json
import time
facts = []
for i in range(6):
  req = http.request('GET', url, headers = headers)
  data = json.loads(req.data.decode('utf_8'))
  facts.append({data["fact"]})
  time.sleep(0.5)
```

```python
data
```

```python
for i in range(len(facts)):
  print(f"{i+1}😂:{facts[i]}")
```
