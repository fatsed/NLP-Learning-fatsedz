```python
from urllib.request import urlretrieve
import zipfile
import pandas as pd

url = "https://archive.ics.uci.edu/static/public/320/student+performance.zip"
urlretrieve(url, "student_performance.zip")
```

```python
with zipfile.ZipFile("student_performance.zip", "r") as z:
    z.extractall("student_performance")
# 3. inner ZIP  =>>>> student.zip
inner_zip = "student_performance/student.zip"
with zipfile.ZipFile(inner_zip, "r") as z:
    z.extractall("student_performance/files")
```

```python
#read file
df_por = pd.read_csv(
    "student_performance/files/student-por.csv",
    sep=";",      # مهم!
)
```
